```python
#Equilibrium conversion checker
import math

raw_kc = input("Enter Kc: ")
print("Raw input type:", type(raw_kc))

try:
    Kc = float(raw_kc)
except ValueError:
    print("Invalid input: Kc must be a number.")
else:
    print("Converted value type:", type(Kc))

    if not math.isfinite(Kc):
        print("Invalid input: Kc must be a finite number.")
    elif Kc <= 0:
        print("Invalid input: Kc must be greater than 0.")
    else:
        # Kc*x^2 - (2.5*Kc + 1)*x + 1.5*Kc = 0
        a = Kc
        b = -(2.5 * Kc + 1)
        c = 1.5 * Kc

        discriminant = b**2 - 4 * a * c

        print(f"Quadratic coefficients: a={a}, b={b}, c={c}")
        print(f"Discriminant: {discriminant}")

        if discriminant < 0:
            print("There are no real roots.")
        else:
            sqrt_discriminant = math.sqrt(discriminant)
            root1 = (-b + sqrt_discriminant) / (2 * a)
            root2 = (-b - sqrt_discriminant) / (2 * a)

            print(f"Root 1: {root1}")
            print(f"Root 2: {root2}")

            valid_roots = [root for root in (root1, root2) if 0 <= root <= 1]

            if not valid_roots:
                print("No physically meaningful root in the range 0 <= x <= 1.")
            else:
                x = valid_roots[0]

                A_eq = 1.5 - x
                B_eq = 1.0 - x
                C_eq = x
                conversion = (x / 1.0) * 100

                print(f"Equilibrium [A]: {A_eq:.3f} mol/L")
                print(f"Equilibrium [B]: {B_eq:.3f} mol/L")
                print(f"Equilibrium [C]: {C_eq:.3f} mol/L")
                print(f"Conversion of B: {conversion:.3f}%")

                if conversion >= 70:
                    print("Conversion classification: high")
                elif conversion >= 40:
                    print("Conversion classification: Moderate")
                else:
                    print("Conversion classification: LOw")
```

# Batch Reactor Simulation and Distillation Code Analysis

Write Python programs for Parts A and B below.

## Part A: Batch Reactor Simulation

A reactant A is consumed in a batch reactor by a first-order reaction:

    dC/dt = -kC

Use Euler’s method with the update equation:

    C_new = C_old - k * C_old * dt

Use the initial concentration C0 = 2.5 mol/L, rate constant k = 0.04 min^-1, and time step dt = 1 min.

1. Use a for loop to simulate the reaction from 0 to 12 minutes. Print a table with time (min), concentration (mol/L), and conversion (%), where conversion is calculated as (C0 - C) / C0 * 100.
2. Use math.exp() to calculate the exact concentration at 12 minutes using C = C0 * exp(-k * t). Print the absolute error between the Euler result and the exact result at 12 minutes.
3. Starting from the initial concentration, use a while loop to simulate the reaction until conversion reaches at least 40% or time reaches 100 minutes. Print the elapsed time and final concentration.

## Part B: Distillation Column Code Analysis

Distillation column sample codes are stored as text, for example, DIST2026X9Q3.

1. Read a code using input().
2. Loop through each character in the code. Count the digits and calculate their sum.
3. Print the number of digits and the sum of the digits.
4. Print a clear message if the code is empty or contains any non-digit characters. For codes containing both letters and digits, still count and sum the digits.

Test Part B with:
- DIST2026X9Q3
- an empty input
- ABCDEF

```python
# Batch reactor simulation and batch tracking 
#part A: 
import math

C0 = 2.5
k = 0.04
dt = 1.0

# Euler simulation from 0 to 12 minutes
concentration = C0

print("Time (min)\tConcentration (mol/L)\tConversion (%)")
for time in range(13):
    conversion = (C0 - concentration) / C0 * 100
    print(f"{time}\t\t{concentration:.4f}\t\t\t{conversion:.2f}")

    if time < 12:
        concentration = concentration - k * concentration * dt

# Compare the Euler result at 12 minutes with the exact result
exact_concentration = C0 * math.exp(-k * 12)
absolute_error = abs(concentration - exact_concentration)

print(f"\nExact concentration at 12 min: {exact_concentration:.6f} mol/L")
print(f"Absolute error at 12 min: {absolute_error:.6f} mol/L")

# Euler simulation until conversion reaches 40% or time reaches 100 minutes
concentration = C0
time = 0
conversion = 0

while conversion < 40 and time < 100:
    concentration = concentration - k * concentration * dt
    time += 1
    conversion = (C0 - concentration) / C0 * 100

print(f"\nTime to reach 40% conversion (or 100 min): {time} min")
print(f"Final concentration: {concentration:.4f} mol/L")
print(f"Final conversion: {conversion:.2f}%")
```

```python
#part B:
code = input("Enter the distillation code: ")

if code == "":
    print("The code is empty.")

digit_count = 0
digit_sum = 0
has_non_digit = False

for character in code:
    if "0" <= character <= "9":
        digit_count += 1
        digit_sum += int(character)
    else:
        has_non_digit = True

if has_non_digit:
    print("The code contains non-digit characters.")

print(f"Number of digits: {digit_count}")
print(f"Sum of digits: {digit_sum}")
```

# Solvent Combinations and Fibonacci Sampling Schedule

Write a Python program with two parts.

## Part A: Solvent Combinations

A laboratory has n candidate solvents and wants to test mixtures containing r solvents. The order of solvents in a mixture does not matter.

1. Write a recursive `factorial(n)` function with the correct base case for 0.
2. Write a `combination(n, r)` function using:
   
       C(n, r) = n! / (r! * (n - r)!)

3. Reject invalid inputs: n or r must be non-negative integers, and r must not be greater than n. Display a clear error message for invalid input.
4. Calculate and print the number of possible mixtures for n=9 and r=3.
5. Test n=6 and r=8 and display the invalid-input message.

## Part B: Fibonacci Sampling Schedule

During a crystallization run, samples are collected at times equal to 4 times the Fibonacci number for that sample.

1. Write a recursive `fib_rec(k)` function that returns the kth Fibonacci number, using F0=0 and F1=1. Reject negative indices.
2. Write an iterative Fibonacci function and use it to check that its results agree with `fib_rec()` for indices 0 through 10.
3. For 10 samples, print a table with sample number, Fibonacci tag (F0, F1, ...), and sampling time in minutes. Calculate each sampling time as the corresponding Fibonacci number multiplied by 4.

```python
#Solvent screening ND SAMPLING SCHEDULE 
# Part A: Solvent combinations

def factorial(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("Factorial requires a non-negative integer.")
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if not isinstance(n, int) or not isinstance(r, int):
        raise ValueError("n and r must be integers.")
    if n < 0 or r < 0 or r > n:
        raise ValueError("Invalid input: require n >= 0 and 0 <= r <= n.")
    return factorial(n) // (factorial(r) * factorial(n - r))


print("Number of mixtures for n=9, r=3:", combination(9, 3))

try:
    print("Number of mixtures for n=6, r=8:", combination(6, 8))
except ValueError as error:
    print(error)


# Part B: Fibonacci sampling schedule

def fib_rec(k):
    if not isinstance(k, int) or k < 0:
        raise ValueError("Fibonacci index must be a non-negative integer.")
    if k <= 1:
        return k
    return fib_rec(k - 1) + fib_rec(k - 2)


# Iterative version used to check the recursive function
def fib_iter(k):
    previous, current = 0, 1
    for _ in range(k):
        previous, current = current, previous + current
    return previous


num_samples = 10

# Check both functions for indices 0 through 10
for k in range(num_samples + 1):
    if fib_rec(k) != fib_iter(k):
        print(f"Fibonacci check failed at index {k}.")
        break
else:
    print(f"Fibonacci check passed: both functions give F10 = {fib_rec(10)}.")

print("\nSample no | Fibonacci tag | Sampling time (min)")
for sample_no in range(1, num_samples + 1):
    fibonacci_index = sample_no - 1
    fibonacci_value = fib_rec(fibonacci_index)
    sampling_time = fibonacci_value * 4
    print(f"{sample_no:9} | F{fibonacci_index:<13} | {sampling_time}")
```

# Kinetic and Gas Data Analysis with Pandas and Matplotlib

Write a Python program using pandas, NumPy, and Matplotlib to analyze the following data.

## Part A: Reaction Kinetics

The reaction concentration was measured at different times for two catalyst conditions:

| Time (min) | Run 1: low catalyst (mol/L) | Run 2: high catalyst (mol/L) |
|---:|---:|---:|
| 0  | 1.50 | 1.50 |
| 10 | 1.17 | 0.96 |
| 20 | 0.91 | 0.61 |
| 30 | 0.71 | 0.39 |
| 40 | 0.55 | 0.25 |
| 50 | 0.43 | 0.16 |

1. Create a pandas DataFrame from the table and save it as `kinetics.json`.
2. Read the saved JSON file back into a DataFrame. Print its shape and the data types of its columns.
3. Print the mean and sample standard deviation of each run. State which run has greater variation based on its standard deviation.
4. Plot both runs on the same line graph, with time on the x-axis. Include a title, axis labels with units, a legend, and a grid.

## Part B: Ethylene Gas Pressure-Volume Data

Ethylene gas is compressed slowly in a piston-cylinder at constant temperature. The measurements are:

| Volume (L) | Pressure (bar) |
|---:|---:|
| 1 | 5.03 |
| 2 | 2.52 |
| 3 | 1.65 |
| 4 | 1.26 |
| 5 | 1.01 |

1. Create a DataFrame from the measurements.
2. Calculate k as the mean of P * V.
3. Plot the measured pressure against volume as a scatter plot. On the same graph, draw the reference curve P = k / V.
4. Add a title, axis labels with units, a legend, and a grid.

```python
#kinetic and gas data : pandas and matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Part A: Kinetic data
kinetics = pd.DataFrame({
    "Time_min": [0, 10, 20, 30, 40, 50],
    "Run1_mol_L_low_catalyst": [1.50, 1.17, 0.91, 0.71, 0.55, 0.43],
    "Run2_mol_L_high_catalyst": [1.50, 0.96, 0.61, 0.39, 0.25, 0.16],
})

kinetics.to_json("kinetics.json", orient="records", indent=2)

# Read the saved JSON back and inspect it
kinetics_loaded = pd.read_json("kinetics.json")
print("DataFrame shape:", kinetics_loaded.shape)
print("\nData types:")
print(kinetics_loaded.dtypes)

# Calculate mean and sample standard deviation
run1 = kinetics_loaded["Run1_mol_L_low_catalyst"]
run2 = kinetics_loaded["Run2_mol_L_high_catalyst"]

print("\nRun 1 mean:", run1.mean())
print("Run 1 standard deviation:", run1.std())
print("Run 2 mean:", run2.mean())
print("Run 2 standard deviation:", run2.std())

if run1.std() > run2.std():
    print("Run 1 shows more variation based on standard deviation.")
else:
    print("Run 2 shows more variation based on standard deviation.")

# Plot both kinetic runs
plt.figure(figsize=(8, 5))
plt.plot(
    kinetics_loaded["Time_min"], run1,
    marker="o", label="Run 1: low catalyst"
)
plt.plot(
    kinetics_loaded["Time_min"], run2,
    marker="s", label="Run 2: high catalyst"
)
plt.title("Reaction Kinetics at Different Catalyst Levels")
plt.xlabel("Time (min)")
plt.ylabel("Concentration (mol/L)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# Part D: Ethylene gas pressure-volume data
gas_data = pd.DataFrame({
    "Volume_L": [1, 2, 3, 4, 5],
    "Pressure_bar": [5.03, 2.52, 1.65, 1.26, 1.01],
})

gas_data["PV"] = gas_data["Pressure_bar"] * gas_data["Volume_L"]
k = gas_data["PV"].mean()
print(f"\nMean value of P*V (k): {k:.3f} bar·L")

# Plot measured pressures and the reference curve P = k / V
volume_curve = np.linspace(
    gas_data["Volume_L"].min(),
    gas_data["Volume_L"].max(),
    200
)

plt.figure(figsize=(8, 5))
plt.scatter(
    gas_data["Volume_L"], gas_data["Pressure_bar"],
    label="Measured data"
)
plt.plot(
    volume_curve, k / volume_curve,
    color="red", label=f"Reference curve: P = k/V (k={k:.3f} bar·L)"
)
plt.title("Ethylene Gas Pressure vs. Volume")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
